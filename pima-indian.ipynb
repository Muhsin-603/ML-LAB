```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
df = pd.read_csv("diabetes.csv")
print(df.head())
```

```python
x, y = df.iloc[:, :-1], df.iloc[:, -1]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model_noscale = LogisticRegression(max_iter=1000)
model_noscale.fit(x_train, y_train)
y_pred_noscale = model_noscale.predict(x_test)
```

```python
metrics_noscale = {
    "Accuracy": accuracy_score(y_test, y_pred_noscale),
    "Recall": recall_score(y_test, y_pred_noscale),
    "Precision": precision_score(y_test, y_pred_noscale),
    "f1 score": f1_score(y_test, y_pred_noscale)
}
```

```python
print("Metrics with no scale : \n")
print(metrics_noscale)
```

```python
scale = StandardScaler()
x_train_scale = scale.fit_transform(x_train)
x_test_scale = scale.fit_transform(x_test)
model_scale = LogisticRegression(max_iter=1000)
model_scale.fit(x_train_scale, y_train)
y_pred_scale = model_scale.predict(x_test_scale)
```

```python
metrics_scale = {
    "Accuracy": accuracy_score(y_test, y_pred_scale),
    "Recall": recall_score(y_test, y_pred_scale),
    "Precision": precision_score(y_test, y_pred_scale),
    "f1 score": f1_score(y_test, y_pred_scale)
}
```

```python
print("With scaling : \n")
print(metrics_scale)
```
