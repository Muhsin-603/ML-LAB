```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
train_df = pd.read_csv("datasets/fashion-mnist_train.csv")
test_df = pd.read_csv("datasets/fashion-mnist_test.csv")
print("Training Data : \n", train_df.head(5), "\n\n")
print("Testing Data : \n", test_df.head(5))
```

```python
x_train = train_df.iloc[:, 1:].values
y_train = train_df.iloc[:, 0].values
x_test = test_df.iloc[:, 1:].values
y_test = test_df.iloc[:, 0].values
```

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
k_val = [1,3,5]
```

```python
for k in k_val:
    print("======== k = ", k, "========")
    knn = KNeighborsClassifier(n_neighbors=k, n_jobs=-1)
    knn.fit(x_train[:10000], y_train[:10000])
    y_pred = knn.predict(x_test[:2000])
    print("Accuracy : ", accuracy_score(y_test[:2000], y_pred))
    print("Classification Report : \n\n", classification_report(y_test[:2000], y_pred))
```

```python
for i in range(5):
    plt.figure(figsize=(2,2))
    plt.imshow(x_test[i].reshape(28,28), cmap='grey', interpolation='nearest')
    plt.title(f"True : {y_test[i]}")
    plt.axis('off')
    plt.show()
```
